```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import plot_tree
```

```python
# Creating the dataset (same dataset used in ID3 example)

data = {
'Weather':['Sunny','Sunny','Cloudy','Rain','Rain','Rain','Cloudy','Sunny','Sunny','Rain','Sunny','Cloudy','Cloudy','Rain'],

'Temperature':['Hot','Hot','Hot','Mild','Cool','Cool','Cool','Mild','Cool','Mild','Mild','Mild','Hot','Mild'],

'Humidity':['High','High','High','High','Normal','Normal','Normal','High','Normal','Normal','Normal','High','Normal','High'],

'Wind':['Weak','Strong','Weak','Weak','Weak','Strong','Strong','Weak','Weak','Weak','Strong','Strong','Weak','Strong'],

'PlayFootball':['No','No','Yes','Yes','Yes','No','Yes','No','Yes','Yes','Yes','Yes','Yes','No']
}

# Convert dictionary to DataFrame
df = pd.DataFrame(data)

# Display dataset
df
```

```python
df.head()
```

```python
le=LabelEncoder()
for col in df.columns:
    df[col]=le.fit_transform(df[col])
df
```

```python
# X = input features
X = df.drop('PlayFootball', axis=1)

# y = target variable
y = df['PlayFootball']

# Display features
print("Features (X):")
display(X)

# Display target
print("Target (y):")
display(y)
```

```python
# Create Decision Tree model using entropy
# entropy criterion = ID3 algorithm

model = DecisionTreeClassifier(criterion='entropy')

# Train the model
model.fit(X, y)

print("Model trained successfully")
```

```python
# Example test data
sample = [[2, 0, 0, 1]]

# Predict result
prediction = model.predict(sample)

print("Prediction:", prediction)
```

```python
# Create a large figure for clear visualization
plt.figure(figsize=(18,10))

# Plot decision tree
plot_tree(
    model,
    feature_names=X.columns,
    class_names=['No','Yes'],
    filled=True,
    rounded=True,
    fontsize=12
)

plt.title("Decision Tree using ID3 Algorithm")
plt.show()
```
